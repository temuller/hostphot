# SED Plotting

With HostPhot, the user can plot the SED from the calculated photometry. This example uses the photometry of SN 2004eo from the [basic example](hostphot_basic_example.ipynb). The photometry of all the surveys found in the object's directory is plotted (they can be selected with `include` and `exclude`).

```python
import hostphot
print('HostPhot version:', hostphot.__version__)
```

```python
from hostphot.photometry.sed_plotting import plot_sed

plot_sed("SN2004eo", phot_type="global")
```

With `plot_flux=True`, the flux densities (in µJy) are plotted instead of the AB magnitudes; non-detections are shown as 3-sigma upper limits:

```python
plot_sed("SN2004eo", phot_type="global", plot_flux=True)
```

For the local photometry, the redshift and the aperture radius (in kpc by default, or arcsec with `ap_units='arcsec'`) are also needed:

```python
plot_sed("SN2004eo", phot_type="local", z=0.0157, radius=3)
```
